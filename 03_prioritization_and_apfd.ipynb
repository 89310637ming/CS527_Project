# 03 Prioritization and APFD

This notebook runs the baseline methods and the proposed method, then computes APFD for each `Lang` bug in the pilot.


```python
import pandas as pd

from d4j_lang_utils import PROJECT_ROOT, run_all_methods

DATA_DIR = PROJECT_ROOT / 'data'
features = pd.read_csv(DATA_DIR / 'lang_test_instances_features.csv')
features.head()
```

```python
ranked_tests, apfd_results = run_all_methods(features)

ranked_tests.head(10)
```

```python
apfd_results.sort_values(['bug_id', 'method']).reset_index(drop=True)
```

```python
ranked_tests.to_csv(DATA_DIR / 'lang_ranked_tests.csv', index=False)
apfd_results.to_csv(DATA_DIR / 'lang_apfd_results.csv', index=False)

print(f'Saved {len(ranked_tests)} ranked rows')
print(f'Saved {len(apfd_results)} APFD rows')
```
