# 02 Feature Extraction

This notebook computes the three pilot factors from the dataset assembled in Notebook 1.

## Current lightweight definitions

1. `failure_history_raw`: prior triggering rate of the test class in earlier pilot bugs
2. `change_frequency_raw`: how often the associated source classes were modified in earlier pilot bugs
3. `complexity_raw`: proxy based on source patch size and hunk count


```python
import pandas as pd

from d4j_lang_utils import PROJECT_ROOT, compute_failure_history, compute_change_frequency, compute_complexity_proxy, normalize_within_bug, compute_risk_score

DATA_DIR = PROJECT_ROOT / 'data'
test_instances = pd.read_csv(DATA_DIR / 'lang_test_instances.csv')
test_instances.head()
```

```python
features = compute_failure_history(test_instances)
features = compute_change_frequency(features)
features = compute_complexity_proxy(features)
features = normalize_within_bug(features)
features = compute_risk_score(features)

features.head(10)
```

```python
columns_to_inspect = [
    'bug_id',
    'test_id',
    'is_triggering',
    'failure_history_raw',
    'change_frequency_raw',
    'complexity_raw',
    'failure_history_norm',
    'change_frequency_norm',
    'complexity_norm',
    'risk_score',
]
features[columns_to_inspect].head(20)
```

```python
features.to_csv(DATA_DIR / 'lang_test_instances_features.csv', index=False)
print(f'Saved {len(features)} feature rows')
```
