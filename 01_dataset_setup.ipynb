# 01 Dataset Setup

This notebook builds the `Lang` pilot dataset directly from the local `Defects4J` checkout in this folder.

## What it produces

- `data/lang_bug_metadata.csv`
- `data/lang_test_instances.csv`

## Important note

For the lightweight pilot, the unit is a **test class** rather than an individual JUnit test method, because `Defects4J` gives `relevant_tests` at the class level.


```python
from pathlib import Path
import pandas as pd

from d4j_lang_utils import PROJECT_ROOT, load_pilot_bug_ids, build_bug_metadata, build_test_instances

DATA_DIR = PROJECT_ROOT / 'data'
DATA_DIR.mkdir(exist_ok=True)

pilot_bug_ids = load_pilot_bug_ids()
pilot_bug_ids
```

```python
bug_metadata = build_bug_metadata(pilot_bug_ids)
bug_metadata[['bug_id', 'revision_buggy', 'revision_fixed', 'src_patch_total_changed', 'src_patch_total_files']]
```

```python
test_instances = build_test_instances(bug_metadata)
test_instances.head(10)
```

```python
bug_metadata.to_csv(DATA_DIR / 'lang_bug_metadata.csv', index=False)
test_instances.to_csv(DATA_DIR / 'lang_test_instances.csv', index=False)

print(f'Saved {len(bug_metadata)} bug rows')
print(f'Saved {len(test_instances)} test-instance rows')
```

```python
validation = pd.DataFrame(
    {
        'bug_id': sorted(test_instances['bug_id'].unique()),
        'num_tests': test_instances.groupby('bug_id')['test_id'].size().sort_index().tolist(),
        'num_triggering_classes': test_instances.groupby('bug_id')['is_triggering'].sum().sort_index().tolist(),
    }
)
validation
```
